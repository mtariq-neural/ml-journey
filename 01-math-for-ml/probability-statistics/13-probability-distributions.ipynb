# Probability Distribution Functions

Before jumping into actual distributions (normal, binomial, etc.), we need to understand a few core terms: random variables, their types, and eventually probability distribution functions themselves. This notebook builds that foundation step by step.

## What is a Random Variable?

A random variable is a variable whose value depends on the outcome of some random process or experiment. It's not "random" in the sense of being unpredictable chaos, it just means we don't know the exact value in advance, but we can describe the possible values it can take and how likely each one is.

For example:
* Flipping a coin and letting X = 1 for heads, 0 for tails. X is a random variable.
* Rolling a die and letting X = the number that shows up. X is a random variable.
* Picking a random customer and recording their `total_bill`. That's also a random variable, drawn from real data instead of a simple game.

The key idea: a random variable is really just a way of mapping outcomes of some random process to numbers, so we can apply math and statistics to them.

## Types of Random Variables

Random variables come in two main types, and this split matters a lot for which kind of distribution or graph we use later.

### 1. Discrete Random Variable
Takes a countable number of distinct values, usually whole numbers. You can list them out: 0, 1, 2, 3...

Examples: number of heads in 3 coin flips, number of customers who walk into a store in an hour, result of rolling a die.

### 2. Continuous Random Variable
Takes any value within a range, including decimals, there are infinitely many possible values, and you cannot list them one by one.

Examples: height of a person, time taken to deliver food, `total_bill` in our tips dataset.

This distinction matters because discrete variables use a **Probability Mass Function (PMF)**, while continuous variables use a **Probability Density Function (PDF)**, which we'll get to shortly.

## What is a Probability Distribution?

A probability distribution is simply a list of all possible outcomes of a random variable, paired with the probability of each outcome happening.

For a discrete random variable, this is easy to picture as a table: each outcome on one side, its probability on the other, and all the probabilities add up to 1.

**Example:** rolling a fair six-sided die. The random variable X can be 1, 2, 3, 4, 5, or 6, and each one has an equal probability of 1/6.

| Outcome (X) | Probability |
| :--- | :--- |
| 1 | 1/6 |
| 2 | 1/6 |
| 3 | 1/6 |
| 4 | 1/6 |
| 5 | 1/6 |
| 6 | 1/6 |

This simple table is already a probability distribution. Once we move to code, we'll see this same idea expressed as a graph instead of a table, which becomes especially useful for continuous variables where you can't list every outcome individually.

## From a Table to a Function

So far, writing a probability distribution as a table worked fine for one die. There are only 6 outcomes, so a 6 row table is easy to write.

But what happens when we roll two dice together and look at their sum? Let's find out.

![Distribution table of two dices](./assets/pdf-two-dices/probability-for-rolling-two-dice.jpg)

```python
import itertools
import pandas as pd

# all possible outcomes when rolling two dice
outcomes = list(itertools.product(range(1, 7), range(1, 7)))
sums = [a + b for a, b in outcomes]

sum_counts = pd.Series(sums).value_counts().sort_index()
sum_probs = sum_counts / len(sums)

sum_probs
```

Writing this table out only took one small table with 11 rows. Still manageable.

But now think bigger. What if we rolled three dice? Four dice? Ten dice? The number of possible sums keeps growing, and the table keeps getting longer and messier to write by hand.

And it gets worse. What if the random variable is not even something countable, like the height of a person? Height can be 170.2 cm, 170.23 cm, 170.238 cm, and so on forever. There is no way to list every single possible value in a table. The table approach completely breaks down here. So we need a function that takes an input and calculate its output, that function is known as probability distribution function.
