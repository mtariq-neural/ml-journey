# What is Feature Engineering?

Feature engineering is the process of using domain knowledge and data understanding to create, change, or select the input columns (features) that a machine learning model learns from.

A model can only learn from what we give it. Good features make the pattern easy to find. Bad or messy features hide it. That is why people often say better features beat a fancier model.

## Why Does It Matter?

Raw data is rarely ready for a model. Columns can be on different scales, categories can be stored as text, values can be missing, and dates can be stuck in a format a model can't use.

Feature engineering fixes these problems, so the model gets clean, useful inputs. In most real projects, this step improves results more than switching to a different algorithm.

## Main Parts of Feature Engineering

Feature engineering has four main parts:

1. **Feature transformation:** changing existing columns into a better form so the model can use them. This includes handling missing values, encoding categories, scaling, and fixing skewed data.
2. **Feature construction:** creating new columns from existing ones. For example, making a `price_per_person` column from `total_bill` and `size`.
3. **Feature selection:** keeping only the columns that help and dropping the ones that don't.
4. **Feature extraction:** combining many columns into fewer, more useful ones. PCA is the common example.

## Feature Transformation

Feature transformation means changing the existing columns into a form the model can use better. It covers these common tasks:

* **Missing values:** filling or removing empty cells.
* **Handling categorical data:** converting text categories into numbers (ordinal, label, and one hot encoding).
* **Outlier detection:** finding values far outside the normal range, like we did with the IQR rule.
* **Feature scaling:** putting columns on a similar scale (standardization and normalization).
* **Distribution fixes:** using log or power transforms on skewed columns.

## Feature Construction

Sometimes the best feature isn't in the data yet, we have to build it. We make new columns by combining or splitting existing ones.

For the tips dataset, a new feature could be the tip as a percentage of the bill. That one number can say more about customer behavior than `tip` or `total_bill` alone.

```python
import seaborn as sns

tips = sns.load_dataset("tips")

tips['tip_percent'] = (tips['tip'] / tips['total_bill']) * 100
tips[['total_bill', 'tip', 'tip_percent']].head()
```

## Feature Selection and Feature Extraction

**Feature selection** means picking the columns that actually help the model. Extra columns can add noise, slow training, and cause overfitting, so dropping useless ones often improves results.

**Feature extraction** means squeezing many columns into fewer new ones that keep most of the information. For example, PCA turns 100 correlated columns into maybe 10 new ones. We will cover this later in the playlist.

## Summary

Feature engineering is about giving the model the best possible inputs. The four parts are transformation, construction, selection, and extraction.
